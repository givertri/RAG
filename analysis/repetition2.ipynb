```python
import pandas as pd
```

```python
# Load the 3 evaluation runs
df1 = pd.read_csv("../eval_results/eval_20260510_125711_reduced.csv")  # seed 42
df2 = pd.read_csv("../eval_results/eval_20260510_140530_reduced.csv")  # seed 43
df3 = pd.read_csv("../eval_results/eval_20260516_122253_reduced.csv")  # seed 44

runs = [df1, df2, df3]

numeric_cols = [
    "response_time",
    "faithfulness",
    "context_precision",
    "constraint_satisfaction",
    "hallucination_rate"
]

# Convert numeric columns for each dataframe
for df in runs:
    df[numeric_cols] = df[numeric_cols].apply(
        pd.to_numeric,
        errors="coerce"
    )

# Append all raw data
all_data = pd.concat(runs, ignore_index=True)

# Group by system and compute mean + stddev
summary = (
    all_data
    .groupby("system")[numeric_cols]
    .agg(["mean", "std"])
)

print(summary)
```
