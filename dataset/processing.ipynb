```python
import pandas as pd
import numpy as np
import chardet
import ast
```

```python
with open("dataset.csv", "rb") as f:
    result = chardet.detect(f.read(100_000))

result

df = pd.read_csv("dataset.csv", encoding=result["encoding"], sep=";")
df.head()
```

```python
df = df.drop(labels=["Author", "Date", "Number of steps", "URL", "Ingredients", "Rating Count"], axis=1)
df
```

```python
df['time'] = df['Preparation Time'] + df['Cooking Time']
df
```

```python
def time_class(minutes):
    if minutes <= 10:
        return "very short"
    elif minutes <= 25:
        return "short"
    elif minutes <= 45:
        return "average"
    elif minutes <= 90:
        return "long"
    else:
        return "very long"

df["time_class"] = df["time"].apply(time_class)
df = df.drop(labels=["Preparation Time", "Cooking Time"], axis=1)
```

```python
df = df.rename(columns={"Rating Value": "rating"})
df = df.rename(columns={"Ingredients_Raw": "ingredients"})
df = df.rename(columns={"Cooking Methods": "methods"})
df = df.rename(columns={"Implements": "utensils"})
```

```python
def safe_literal_eval(x):
    # 1. Handle NaN / None early
    if pd.isna(x):
        return None

    # 2. Ensure it's a string
    if not isinstance(x, str):
        return None

    x = x.strip()

    # 3. Attempt to fix common broken dict issue (missing closing brace)
    if x.startswith("{") and not x.endswith("}"):
        x = x + "}"

    try:
        return ast.literal_eval(x)
    except Exception as e:
        print("⚠️ Failed to parse value:")
        print(x)
        print("Error:", e)
        return None

df["Nutrition"] = df["Nutrition"].apply(safe_literal_eval)
nutrition_df = pd.json_normalize(df["Nutrition"])

nutrition_df.columns = (
    nutrition_df.columns
    .str.lower()
    .str.replace(" ", "_", regex=False)
)

unit_mapping = {
    "calories": "kcal",
    "carbohydrates": "g",
    "cholesterol": "mg",
    "fiber": "g",
    "protein": "g",
    "saturated_fat": "g",
    "sodium": "mg",
    "sugar": "g",
    "fat": "g",
    "unsaturated_fat": "g"
}

nutrition_df.rename(columns={col: f"{col}_{unit}" for col, unit in unit_mapping.items() if col in nutrition_df.columns}, inplace=True)

df = df.drop(columns="Nutrition").join(nutrition_df)

nutrition_cols = [
    "calories_kcal",
    "carbohydrates_g",
    "cholesterol_mg",
    "fiber_g",
    "protein_g",
    "saturated_fat_g",
    "sodium_mg",
    "sugar_g",
    "fat_g",
    "unsaturated_fat_g"
]

for col in nutrition_cols:
    if df[col].dtype == object:
        df[col] = (
            df[col]
            .astype(str)
            .str.extract(r"([\d\.]+)", expand=False)
            .pipe(pd.to_numeric, errors="coerce")
        )
```
